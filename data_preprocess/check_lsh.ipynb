```python
import os
import time
import argparse
from tqdm import tqdm
from gensim.utils import simple_preprocess  
from simhash import Simhash, SimhashIndex


data_path = '/data1/redpajama-v2-process/data_preprocess/2014-15.txt'
threshold, f = 3, 128

# data, hashs = [], []
# n, num = 0, 100000
# with open(data_path, 'r') as f:
#     para = ''
#     for line in tqdm(f):
#         if line == '[end of paragraph --wyc]\n':
#             data.append(para)
#             hashs.append(Simhash(simple_preprocess(para), f=128))
#             para = ''
#             n += 1
#             if n == num:
#                 break
#         else:
#             para += line
        
# print(len(data))


start, end = 0, 100000
example = 1000
data, hashs = [], []
n, n_exp = 0, 0
with open(data_path, 'r') as f:
    para = ''
    for line in tqdm(f):
        if line == '[end of paragraph --wyc]\n':
            n += 1
            if n < start:
                continue
            if n == start:
                print(start)
            if n == end:
                break
            if n % 1000 == 0:
                print(n)
            hash = Simhash(simple_preprocess(para), f=128)
            hashs.append(hash)
            data.append(para)
            para = ''
        else:
            para += line
```

```python
outlier = 0
ratio = []
pairs = []
for i in tqdm(range(len(hashs)-1)):
    hash, para = hashs[i], data[i]
    for j in range(i+1, len(hashs)):
        if hashs[j].distance(hash) <= threshold:
            # print('sim: %d %d'%(i, j))
            # print(len(data[j]), data[j][:100])
            # print('----------------------------------')
            # print(len(para), para[:100])
            # print('----------------------------------------------------------------')
            print(len(data[j]), len(para))
            ratio.append(len(para) / len(data[j]))
            pairs.append((i, j))
            if len(para) / len(data[j]) >= 2 or len(para) / len(data[j]) <= 1/2:
                outlier += 1
            n_exp += 1
            break
        if n_exp == example:
            print(n_exp, outlier)
            assert False
```
